## sales analysis

#### total revenue

```python
total_revenue = d4['payment_value'].sum()
f'Total Revenue : {total_revenue} $'
```

#### average order value

```python
duplicate_cart = d4['order_id'].duplicated().sum()
f'Duplicate cart : {duplicate_cart}'
```

```python
unique_order = d4['order_id'].nunique()
f'unique order : {unique_order}'
```

```python
aov = total_revenue / unique_order
f'average over value : {aov} $'
```

#### monthly sales

```python
merged = pd.merge(d6, d4, on ='order_id')
monthly_sales = (merged.groupby(['order_purchase_timestamp_month'])['payment_value']).sum().reset_index()
monthly_sales.columns = ['month', 'total sales']
monthly_sales
```

#### monthly sales by year

```python
monthly_sales_year = merged.groupby(['order_estimated_delivery_date_year','order_estimated_delivery_date_month'])['payment_value'].sum().reset_index()
pd.set_option('display.float_format', '{:.0f}'.format)
monthly_sales_year.columns = ['year' , 'month' , 'total sale']
monthly_sales_year
```

#### monthly sales by year plot

```python
sns.barplot(
    data=monthly_sales_year,
    x='month',
    y='total sale',
    hue='year'
)
plt.show()
```

## delivery analysis

```python
d6['delivery_days'] = (d6['order_delivered_customer_date']-d6['order_purchase_timestamp']).dt.days
```

```python
d6['order_estimated_delivery_days'] = (d6['order_estimated_delivery_date']-d6['order_purchase_timestamp']).dt.days
```

## new vs repeat customer

```python
order_per_customer = merge5.groupby(['customer_id_x'])['order_item_id'].nunique()
customer_type = np.where(
    order_per_customer == 1,
    'One-time Customer',
    'Repeat Customer'
)
pd.Series(customer_type).value_counts()
```

```python
repeat_rate = ((order_per_customer > 1).mean() * 100)

f'Repeat Customer Rate: {repeat_rate:.2f}%'
```

## import to postgresql

```python
from sqlalchemy import create_engine

username = 'postgres'
password = '123456'
host = 'localhost'
port = '5432'
database = 'olist'

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")
table_name1 = 'customer'
table_name2 = 'geolocation'
table_name3 = 'orderitem'
table_name4 = 'payment'
table_name5 = 'review'
table_name6 = 'date'
table_name7 = 'product'
table_name8 = 'seller'

d1.to_sql(table_name1, engine, if_exists='replace', index=False)
d2.to_sql(table_name2, engine, if_exists='replace', index=False)
d3.to_sql(table_name3, engine, if_exists='replace', index=False)
d4.to_sql(table_name4, engine, if_exists='replace', index=False)
d5.to_sql(table_name5, engine, if_exists='replace', index=False)
d6.to_sql(table_name6, engine, if_exists='replace', index=False)
d7.to_sql(table_name7, engine, if_exists='replace', index=False)
d8.to_sql(table_name8, engine, if_exists='replace', index=False)
print(f"Data successfully loaded into tables: {table_name1}, {table_name2}, {table_name3}, {table_name4}, {table_name5}, {table_name6}, {table_name7}, {table_name8} in database '{database}'.")
```
